# Task 3: Exploratory Data Analysis on IPL Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("matches.csv")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
team_wins = df['winner'].value_counts()

team_wins.head(10)
```

```python
plt.figure(figsize=(10, 5))

team_wins.head(10).plot(kind='bar')

plt.title("Top IPL Teams by Number of Wins")
plt.xlabel("Team")
plt.ylabel("Wins")

plt.xticks(rotation=45)
plt.show()
```

```python
df['toss_decision'].value_counts()
```

```python
sns.countplot(data=df, x='toss_decision')

plt.title("Toss Decisions")
plt.show()
```

```python
top_players = df['player_of_match'].value_counts()

top_players.head(10)
```

```python
plt.figure(figsize=(10, 5))

top_players.head(10).plot(kind='bar')

plt.title("Top 10 Players with Most Player of the Match Awards")
plt.xlabel("Player")
plt.ylabel("Awards")

plt.xticks(rotation=45)
plt.show()
```

### Conclusion

The IPL dataset was analyzed to identify winning teams,
toss decisions and players who received the most Player of
the Match awards. Graphs were used to visualize the findings.
