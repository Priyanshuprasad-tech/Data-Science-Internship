```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    'Math': [85, 78, 92, 65, 88, 76, 95, 70, 82, 90],
    'Science': [88, 75, 94, 60, 85, 72, 96, 68, 80, 91],
    'English': [80, 82, 88, 70, 84, 78, 90, 75, 79, 86],
    'Attendance': [95, 90, 98, 75, 92, 85, 99, 80, 88, 96]
}

df = pd.DataFrame(data)

df
```

```python
correlation = df.corr()

correlation
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
